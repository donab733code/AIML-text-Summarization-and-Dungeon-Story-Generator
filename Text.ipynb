```python
import transformers
print(transformers.__version__)
```

```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)
```

```python
text = """
Artificial intelligence is a rapidly growing technology that allows
machines to learn from data and perform tasks that normally require
human intelligence. AI is used in healthcare, education, finance,
transportation, and many other industries.
"""

summary = summarizer(
    text,
    max_length=50,
    min_length=20,
    do_sample=False
)

print(summary[0]["summary_text"])
```

```python
article ="""Artificial intelligence (AI) has become one of the most influential technologies of the 21st century. It enables computers to perform tasks that traditionally required human intelligence, such as recognizing speech, understanding language, solving problems, and making decisions. Over the past decade, advances in machine learning and deep learning have significantly improved the capabilities of AI systems.

In healthcare, AI is helping doctors detect diseases earlier and more accurately. Machine learning models can analyze medical images, identify patterns that may not be visible to the human eye, and assist in diagnosing conditions such as cancer or heart disease. AI-powered systems are also being used to develop personalized treatment plans and accelerate the discovery of new medicines.

The education sector has also benefited from AI. Intelligent tutoring systems can adapt lessons to individual students based on their learning pace and performance. Virtual assistants answer students' questions, while automated grading systems reduce teachers' workloads, allowing them to focus more on instruction and student engagement.

Businesses use AI to improve customer service, optimize supply chains, and make data-driven decisions. Chatbots provide 24-hour customer support, recommendation systems suggest products based on user preferences, and predictive analytics help companies forecast demand and reduce operational costs.

Despite these benefits, AI also presents several challenges. The increasing use of automation may replace certain jobs, requiring workers to learn new skills. Privacy concerns have emerged because AI systems often rely on large amounts of personal data. In addition, biased training data can lead to unfair or inaccurate decisions, highlighting the importance of developing ethical and transparent AI systems.

Governments, researchers, and technology companies are working together to establish regulations and ethical guidelines for AI development. These efforts aim to ensure that AI remains safe, fair, and beneficial to society while encouraging continued innovation. As AI continues to evolve, it is expected to play an even greater role in shaping healthcare, education, business, transportation, and many other aspects of everyday life."""
summary = summarizer(
    article,
    max_length=120,
    min_length=40,
    do_sample=False
)

print(summary[0]["summary_text"])

```
